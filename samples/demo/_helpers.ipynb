# Shared helpers
Pulled into other notebooks with `%run ./_helpers` — everything defined here becomes available there.

```python
from pyspark.sql import functions as F

CURRENCY = "₹"

def inr(x):
    """Format a number as Indian rupees with lakh/crore units."""
    x = float(x or 0)
    if abs(x) >= 1e7: return f"{CURRENCY}{x/1e7:.2f} Cr"
    if abs(x) >= 1e5: return f"{CURRENCY}{x/1e5:.2f} L"
    return f"{CURRENCY}{x:,.0f}"

def add_margin(df):
    """Adds gross_margin = net_sales - cost."""
    return df.withColumn("gross_margin", F.round(F.col("net_sales") - F.col("cost"), 2))

print("helpers loaded: inr(), add_margin(), CURRENCY")
```
