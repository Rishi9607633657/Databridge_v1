# Child notebook: region report
Called by the tour with `dbutils.notebook.run(...)`. It reads its **arguments as widgets** and returns a JSON result with `dbutils.notebook.exit()`.

```python
dbutils.widgets.text("region", "West")
dbutils.widgets.text("table", "demo.sales")
region = dbutils.widgets.get("region")
table = dbutils.widgets.get("table")
print(f"Building the report for region={region} from {table}")
```

```python
import json
from pyspark.sql import functions as F
row = (spark.table(table).where(F.col("region") == region)
       .agg(F.round(F.sum("net_sales"), 2).alias("net_sales"), F.count("*").alias("orders")).first())
result = {"region": region, "net_sales": float(row["net_sales"] or 0), "orders": int(row["orders"])}
print(result)
```

```python
dbutils.notebook.exit(json.dumps(result))
```
